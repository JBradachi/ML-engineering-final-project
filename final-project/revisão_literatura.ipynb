# Revisão da literatura

Identificar insetos não é um problema novo na literatura, mas ainda é um problema em aberto por conta das inumeras espécies.

Até 2020 eram usadas abordagens sem redes neurais, o que gerava resultados bem ruins. A partir desse ano começaram a aplicar a abordagem e refinar com algumas metodologias.

Foquei em procurar artigos que abordam redes neurais somados ao problema de classificação de insetos usando imagens

A revisão da literatura [1] apresenta o problema, uma metodologia abstrata de como é geralmente tratado, mostra alguns datasets comumente usados na literatura (focados em crop pests). Para resolver o problema, é realizada a segmentação das imagens e posteriormente a classificação. Eles discutem sobre o uso de algoritmos de aprendizado de máquina clássicos comparados aos de aprendizado profundo, e é citado que para a maioria dos casos DL é bem melhor. Até a data que o artigo cobre (2024), a maioria dos trabalhos focavam em identificar diferentes tipos de espécies em menor quantidade de classes. Inclusive eles sempre trataram em situações específicas (abordagens focando em uma única família de insetos).

Abordagens usaram ResNet50, LeNet, AlexNet, GoogLeNet, YOLO v5-CB (e outras versões do YOLO na época), MobileNet
